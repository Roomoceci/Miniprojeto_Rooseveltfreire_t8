### Etapa 4 - Limpeza, Padronização e Tratamento da Base

```python
# 1. Remover colunas desnecessárias e totalmente vazias (Unnamed)
colunas_para_remover = ['Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13']
df = df.drop(columns=colunas_para_remover, errors='ignore')
print("Colunas vazias 'Unnamed' removidas com sucesso!")

# 2. Padronização das colunas de texto (Letras maiúsculas e sem espaços extras)
colunas_texto_ajuste = ['CL_GENERO', 'CL_SEG', 'PR_CAT', 'PR_NOME']
for col in colunas_texto_ajuste:
    df[col] = df[col].astype(str).str.strip().str.upper()

# Tratar inconsistência de valores '#N/D' identificados na coluna de categoria do produto
df['PR_CAT'] = df['PR_CAT'].replace('#N/D', 'NAO DEFINIDO')
print("Colunas textuais padronizadas e inconsistências como '#N/D' tratadas!")
```

### Etapa 5 - Organização em Funções (Tratamento e Resumo)

```python
def tratar_dados(dataframe):
    """
    Trata, limpa e padroniza o DataFrame de varejo.
    Retorna o DataFrame limpo.
    """
    # Cria uma cópia para evitar warnings de SettingWithCopy
    df_clean = dataframe.copy()

    # 1. Remover colunas desnecessárias e vazias (Unnamed: 10 a 13)
    colunas_para_remover = [
        'Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13'
        ]
    df_clean = df_clean.drop(columns=colunas_para_remover, errors='ignore')

    # 2. Padronização de textos em colunas específicas
    colunas_texto = [
        'CL_GENERO', 'CL_SEG', 'PR_CAT', 'PR_NOME'
        ]
    for col in colunas_texto:
        if col in df_clean.columns:
            df_clean[col] = df_clean[col].astype(str).str.strip().str.upper()

    # 3. Tratamento de inconsistência textual específica
    if 'PR_CAT' in df_clean.columns:
        df_clean['PR_CAT'] = df_clean['PR_CAT'].replace('#N/D', 'NAO DEFINIDO')

    # 4. Remoção de duplicatas absolutas (linhas idênticas)
    df_clean = df_clean.drop_duplicates()

    return df_clean
```

```python
# Para visualizar o DataFrame tratado após a execução da função, usamos a variável global 'df_tratado'
print("Colunas que restaram após o tratamento:")
print(df_tratado.columns.tolist())

print("\nPrimeiras linhas da base tratada:")
display(df_tratado.head())
```

```python
def gerar_resumo(dataframe):
    """
    Gera e retorna um resumo estatístico e de distribuição das principais colunas do DataFrame.
    """
    total_linhas = dataframe.shape[0]
    total_colunas = dataframe.shape[1]

    # Distribuição da coluna PR_CAT caso exista no DataFrame
    if 'PR_CAT' in dataframe.columns:
        distribuicao_categorias = dataframe['PR_CAT'].value_counts()
    else:
        distribuicao_categorias = "Coluna PR_CAT não encontrada."

    resumo_dict = {
        "Total de Linhas": total_linhas,
        "Total de Colunas": total_colunas,
        "Distribuicao de Categorias": distribuicao_categorias
    }

    return resumo_dict
```

```python
# Testando a aplicação prática das duas funções criadas a partir do df_original

# 1. Aplicando a função de tratamento
df_tratado = tratar_dados(df_original)

# 2. Obtendo o resumo estruturado através da função de resumo
resumo_final = gerar_resumo(df_tratado)

# Exibindo os resultados retornados
print(f"Linhas tratadas: {resumo_final['Total de Linhas']}")
print(f"Colunas restantes: {resumo_final['Total de Colunas']}\n")
print("Distribuição de Categorias resultante:")
display(resumo_final['Distribuicao de Categorias'])
```

### Etapa 7 - Segmentação com Filtros e Análise Agrupada (groupby)

```python
# 1. Aplicando o Filtro: Clientes que possuem ao menos 1 dependente/filho (CL_FHL > 0)
df_familias = df_tratado[df_tratado['CL_FHL'] > 0].copy()

# 2. Primeiro agrupamento: Contagem de registros de itens por Categoria de Produto
resumo_categoria = df_familias.groupby('PR_CAT').size().reset_index(name='Qtd_Registros_Itens')

# 3. Segundo agrupamento: Distribuição de registros de itens por Gênero do Cliente cadastrado
resumo_genero = df_familias.groupby('CL_GENERO').size().reset_index(name='Qtd_Registros_Itens')

print(f"Total de registros na base filtrada (Famílias com filhos): {df_familias.shape[0]}\n")

print(" Agrupamento 1: Registros de itens consumidos por Categoria (Famílias com Filhos) ===")
display(resumo_categoria.sort_values(by='Qtd_Registros_Itens', ascending=False))

print("\n Agrupamento 2: Registros de itens por Gênero do Cliente Cadastrado (Famílias com Filhos) ===")
display(resumo_genero)
```

Explicação das Métricas e Medidas

Para interpretar corretamente estes agrupamentos, é fundamental não confundir as diferentes entidades de nossa base de dados:

 O que cada agrupamento acima mede?
  Agrupamento por Categoria: Mede a quantidade absoluta de registros de itens (linhas na tabela) que foram levados nas cestas destas famílias. Por exemplo, se uma linha indica ALIMENTOS, significa que um item pertencente a alimentos foi adicionado em alguma compra.
  Agrupamento por Gênero: Mede o volume de itens associados a cadastros de clientes daquele gênero específico dentro deste grupo filtrado.

Diferenciando as Entidades para Evitar Confusões:
  Registros de Itens (Linhas): Cada linha representa a inclusão de um produto em uma compra.O número total obtido nas tabelas acima (ex: Qtd_Registros_Itens) representa esse volume de itens faturados, e não o número de compras ou clientes individuais.
  Compras (Transações / CO_ID): Uma única compra (CO_ID) pode conter múltiplos itens (várias linhas no nosso dataset). Para contar o número de transações reais, precisaríamos contar os IDs de compra únicos (.nunique()).
  Clientes (CL_ID): Um cliente cadastrado pode realizar várias compras ao longo do tempo. Para sabermos quantos consumidores únicos com filhos temos na base, deveríamos contar os CL_ID exclusivos.
  Faturamento: Como nossa base não contém colunas de preços ou valores monetários monetizados das compras nesta tabela, não podemos estimar o faturamento financeiro diretamente aqui, apenas o volume físico de movimentação de produtos.
